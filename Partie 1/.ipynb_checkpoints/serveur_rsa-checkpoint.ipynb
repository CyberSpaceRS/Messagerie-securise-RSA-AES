```python
# serveur_rsa.py
from flask import Flask, request, jsonify
import rsa

app = Flask(__name__)

# --------------------------
# Génération de la paire de clés RSA
cle_publique, cle_privee = rsa.newkeys(512)  # 512 bits pour test local (éviter >245 octets)

# --------------------------
# Route pour fournir la clé publique
@app.route('/get_key', methods=['GET'])
def get_key():
    return jsonify({
        'n': cle_publique.n,
        'e': cle_publique.e
    })

# --------------------------
# Route pour recevoir un message chiffré
@app.route('/message', methods=['POST'])
def recevoir_message():
    donnees = request.get_json()
    message_chiffre = bytes.fromhex(donnees['message'])

    try:
        message_clair = rsa.decrypt(message_chiffre, cle_privee)
        return jsonify({ 'message_clair': message_clair.decode() })
    except Exception as e:
        return jsonify({ 'erreur': str(e) }), 400

if __name__ == '__main__':
    app.run(port=5000)
```
