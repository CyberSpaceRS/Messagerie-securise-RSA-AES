```python
import requests
import rsa

# Étape 1 : Récupération de la clé publique depuis le serveur
url_serveur = "http://127.0.0.1:5000"
reponse = requests.get(f"{url_serveur}/get_key")
cle_publique_data = reponse.json()

n = cle_publique_data['n']
e = cle_publique_data['e']
cle_publique = rsa.PublicKey(n, e)

# Étape 2 : Préparation et chiffrement du message
message_original = "Bonjour, ceci est un message RSA."
message_chiffre = rsa.encrypt(message_original.encode(), cle_publique)

# Étape 3 : Envoi du message chiffré
reponse = requests.post(f"{url_serveur}/message", json={
    'message': message_chiffre.hex()
})

# Étape 4 : Affichage de la réponse du serveur
if reponse.status_code == 200:
    print("🔓 Réponse du serveur :", reponse.json()['message_clair'])
else:
    print("❌ Erreur :", reponse.text)
```
